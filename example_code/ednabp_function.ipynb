# Module `SeqAnalyser`

```python
import importlib
import ednabp
importlib.reload(ednabp)
```

```python
from ednabp import SampleData
CJ_data = SampleData()
CJ_data.unpickle_data(pickle_path="./edNA/test_on_data/CJ_test/CJ_data.pkl")
```

```python
from ednabp import SeqAnalyser
seq_analyser = SeqAnalyser(sampledata=CJ_data)
```

```python
import numpy as np
family_list = []
for sample_id in CJ_data.sample_id_list:
    family_list.extend([lv["family"] for lv in CJ_data.sample_data[sample_id].hap2level.values()])
family_list = np.unique(family_list)
```

```python
seq_analyser.write_umap_index(
    taxa_list = family_list,
    taxa_level = "family",
    unit_level = "species",
    save_dir = "./edNA/test_on_data/CJ_test",
    neighbors = 15,
    min_dist = 0.1,
    random_state = None,
    calc_dist = False,
    sample_id_list = None,
)
```

```python
seq_analyser.plot_umap(
    index_path = "./edNA/test_on_data/CJ_test/umap_index.tsv",
    n_unit_threshold = 15,
    category = "all",
    save_dir = "./edNA/test_on_data/CJ_test/",
    cmap = "Spectral",
    show_legend = True
)
```
